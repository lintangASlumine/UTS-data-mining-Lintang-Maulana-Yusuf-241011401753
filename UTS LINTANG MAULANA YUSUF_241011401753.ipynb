```python
inventory = {
    "laptop": {"stok": 10, "harga": 15000000},
    "mouse": {"stok": 50, "harga": 250000},
    "monitor": {"stok": 20, "harga": 3000000}
}

# 1. Tambahkan produk baru
inventory["keyboard"] = {
    "stok": 30,
    "harga": 500000
}

# 2. Perbarui harga laptop
inventory["laptop"].update({
    "harga": 14500000
})

# 3. Hitung total nilai aset
total_aset = 0

for barang in inventory.values():
    total_aset += barang["stok"] * barang["harga"]

# Output
print("Data Inventory:")

for nama, info in inventory.items():
    print(nama)
    print("Stok :", info["stok"])
    print("Harga:", info["harga"])
    print()

print("Total Nilai Aset Gudang:", total_aset)
```

```python
import numpy as np

# Matriks nilai mahasiswa (baris = mahasiswa, kolom = tugas 1,2,3)
scores = np.array([
    [80, 85, 90],
    [70, 60, 75],
    [95, 90, 100],
    [40, 50, 45]
])

# 1. Hitung rata-rata per mahasiswa (baris)
average_scores = np.mean(scores, axis=1)

# 2. Cari nilai tertinggi dari seluruh data
max_score = np.max(scores)

# 3. Buat array hasil kelulusan dengan np.where
statuses = np.where(average_scores >= 70, "Lulus", "Gagal")

# Cetak dalam format tabel
print(f"{'Mahasiswa':<10} {'Tugas 1':<8} {'Tugas 2':<8} {'Tugas 3':<8} {'Rata-Rata':<10} {'Status':<6}")
print('-' * 55)

for i, row in enumerate(scores):
    avg = average_scores[i]
    status = statuses[i]
    print(f"{i+1:<10} {row[0]:<8} {row[1]:<8} {row[2]:<8} {avg:<10.2f} {status:<6}")

print("\nNilai tertinggi dari seluruh data:", max_score)
```

```python
import pandas as pd
import numpy as np

# Data awal
df_data = {
    'Nama': ['Andi', 'Budi', 'Caca', 'Dedi', 'Euis'],
    'Departemen': ['IT', 'HR', 'IT', 'Sales', 'Sales'],
    'Gaji': [8000000, 7000000, 8500000, np.nan, 6000000],
    'Pengalaman_Tahun': [5, 3, 6, 2, 1]
}

# Membuat DataFrame
df = pd.DataFrame(df_data)

# 1. Isi nilai Gaji yang kosong (NaN) dengan median dari seluruh gaji
median_gaji = df['Gaji'].median()
df['Gaji'] = df['Gaji'].fillna(median_gaji)

# 2. Buat kolom baru 'Kategori_Senioritas'
# Ketentuan: Jika pengalaman > 4 tahun maka 'Senior', jika tidak 'Junior'.
df['Kategori_Senioritas'] = df['Pengalaman_Tahun'].apply(
    lambda x: 'Senior' if x > 4 else 'Junior'
)

# 3. Tampilkan rata-rata Gaji per Departemen menggunakan .groupby()
rata_rata_gaji_per_dept = df.groupby('Departemen')['Gaji'].mean()

# Output hasil
print("DataFrame setelah manipulasi:")
print(df)

print("\nRata-rata Gaji per Departemen:")
print(rata_rata_gaji_per_dept)
```
